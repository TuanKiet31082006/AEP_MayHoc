```python
import pandas as pd
import numpy as np
import joblib
import os

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Đọc dữ liệu
df = pd.read_csv("../Data/processed/appliances_processed.csv")

# Tách đặc trưng và biến mục tiêu
X = df.drop(columns=["Appliances","date"])
y = df["Appliances"]

# Chia dữ liệu theo thứ tự thời gian (80/20)
split_index = int(len(X) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

# Nạp mô hình tốt nhất
best_model = joblib.load("../models/best_model.pkl")

print("Đã nạp mô hình thành công!")
print("Số dòng dữ liệu kiểm tra:", len(X_test))
```

```python
# Thực hiện dự đoán
y_pred = best_model.predict(X_test)

# Tạo bảng so sánh
prediction_df = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

# Tính sai số
prediction_df["Error"] = (
    prediction_df["Actual"] - prediction_df["Predicted"]
).abs()

# Hiển thị 10 kết quả đầu tiên
display(prediction_df.head(10))
```

```python
# Tạo bảng kết quả dự đoán
prediction_df = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

# Tính sai số tuyệt đối
prediction_df["Error"] = abs(
    prediction_df["Actual"] - prediction_df["Predicted"]
)

# Hiển thị 10 kết quả đầu tiên
display(prediction_df.head(10))
```

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION")
print("-------------------------")
print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
import os

# Tạo thư mục reports nếu chưa có
os.makedirs("../reports", exist_ok=True)

# Xuất bảng kết quả
prediction_df.to_csv(
    "../reports/prediction_results.csv",
    index=False
)

print("Đã lưu kết quả dự đoán thành công!")
```
